```python
product_info = pd.DataFrame({
    "Product name" :[
        "Laptop",
        "Mouse",
        "Printer",
          "Tablet",
          "Monitor"
         
    ],
    "Supplier":[
        "Dell",
        "Logitech",
         "HP",
        "Samsung",
         "LG"
    ],
    "Discount":[
        10,
        5,
        15,
        8,
        12
    ],
    
})
```

```python
print("13. PRODUCT INFOMATION DATASET")
print("-" *50)
display(product_info)
```

```python
merge_df=pd.merge(
df,
product_info,
on = "Product name",
how = "left"
)
```

```python
print("14. MERGED DATASET")
print("-" *50)
display(merge_df)
```

```python
region_info = pd.DataFrame({
    " Region_Manager" :[
        "Manager A",
        "Manager B",
        "Manager C",
        "Manager D"
    ]
         
        },
   
    index = [
        "North",
        "South",
        "East",
        "West"
    ]
        )
```

```python
print("14. REGION INFOMATION DATASET")
print("-" *50)
display(region_info)
df_region=df.set_index("Region")
joined_df=df_region.join(region_info)
print("15. JOINED INFOMATION DATASET")
print("-" *50)
display(joined_df)
```

```python
region_sales=df.groupby(
"Region"
)["Sales"].sum()
```

```python
print("16. TOTAL SALES BY REGION")
print("-" *50)
display(region_sales)
```

```python
pivot_table=pd.pivot_table(
df,
    values="Sales",
    index="Category",
    columns="Region",
    aggfunc="sum",
    fill_value = 0
)
```
